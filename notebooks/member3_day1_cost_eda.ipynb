# Member 3 - Cost EDA (Day 1)

```
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('../data/processed/paimana_master_v1.csv')
df['cost_change'] = df['revised_cost'] - df['original_cost']
df.shape
```

## 1. Cost Overview

```
print(f"Projects with cost increase: {len(df[df['cost_change'] > 0])}")
print(f"Projects unchanged: {len(df[df['cost_change'] == 0])}")
print(f"Projects with cost decrease: {len(df[df['cost_change'] < 0])}")
```

## 2. Statistical Summary of Overruns

```
df[df['cost_change'] > 0][['cost_change', 'cost_overrun_percent']].describe()
```

## 3. Cost vs Progress Correlation

```
df[['cost_overrun_percent', 'physical_progress', 'expenditure_percent']].corr()
```

## 4. State & Agency Breakdown

```
df.groupby('state')['cost_overrun'].mean().sort_values(ascending=False).head()
```
